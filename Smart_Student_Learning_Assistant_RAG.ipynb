```python
import faiss
import numpy as np

# Create a FAISS index
dimension = 1536
index = faiss.IndexFlatL2(dimension)

print("FAISS initialized successfully!")
print("Vector dimensions:", dimension)
print("Number of stored vectors:", index.ntotal)
```

```python
from google.colab import files

uploaded = files.upload()

pdf_path = list(uploaded.keys())[0]

print("Uploaded file:", pdf_path)
```

```python
from pypdf import PdfReader

# Read the uploaded PDF
reader = PdfReader(pdf_path)

# Extract text from all pages
pdf_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        pdf_text += text + "\n"

# Display results
print("PDF text extracted successfully!")
print("Total pages:", len(reader.pages))
print("Total characters:", len(pdf_text))
print("\nFirst 1000 characters:\n")
print(pdf_text[:1000])
```

```python
# Split extracted text into chunks
chunk_size = 500
overlap = 100

chunks = []

start = 0

while start < len(pdf_text):
    end = start + chunk_size
    chunk = pdf_text[start:end]

    chunks.append(chunk)

    start += chunk_size - overlap

print("Text splitting completed!")
print("Total chunks:", len(chunks))

# Display first chunk
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
# Store embeddings for all chunks
embeddings = []

for chunk in chunks:
    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=chunk
    )

    embedding = response.data[0].embedding
    embeddings.append(embedding)

# Convert embeddings into a NumPy array
embeddings_array = np.array(embeddings, dtype="float32")

print("All chunk embeddings generated successfully!")
print("Embedding shape:", embeddings_array.shape)
```

```python
# Add embeddings to the FAISS index
index.add(embeddings_array)

print("Embeddings stored in FAISS successfully!")
print("Total vectors stored:", index.ntotal)
```

```python
# Student question
query = "What is Artificial Intelligence?"

# Generate embedding for the question
query_response = client.embeddings.create(
    model="text-embedding-3-small",
    input=query
)

query_embedding = np.array(
    [query_response.data[0].embedding],
    dtype="float32"
)

# Search FAISS for the 3 most relevant chunks
distances, indices = index.search(query_embedding, k=3)

print("Student Question:", query)
print("\nMost relevant study material:\n")

for i, idx in enumerate(indices[0]):
    print(f"Result {i+1}:")
    print(chunks[idx])
    print("-" * 50)
```

```python

import gradio as gr
import numpy as np

def student_assistant(question):

    if not question.strip():
        return "Please enter a question."

    # Generate question embedding
    query_response = client.embeddings.create(
        model="text-embedding-3-small",
        input=question
    )

    query_vector = np.array(
        [query_response.data[0].embedding],
        dtype="float32"
    )

    # Retrieve relevant chunks
    distances, indices = index.search(query_vector, k=3)

    # Check relevance
    if distances[0][0] > 1.0:
        return "The requested information is not available in the uploaded study material."

    # Prepare context
    context = "\n\n".join(
        [chunks[idx] for idx in indices[0] if idx >= 0]
    )

    # Generate answer
    response = client.chat.completions.create(
        model="nova-micro",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a Smart Student Learning Assistant. "
                    "Answer ONLY using the provided study material. "
                    "Do not use general knowledge or outside information. "
                    "If the answer is not supported by the study material, "
                    "respond only with: "
                    "'The requested information is not available in the uploaded study material.' "
                    "Do not add outside explanations or examples."
                )
            },
            {
                "role": "user",
                "content": f"""
Study Material:
{context}

Student Question:
{question}
"""
            }
        ],
        max_tokens=300
    )

    return response.choices[0].message.content


# Create Gradio interface
demo = gr.Interface(
    fn=student_assistant,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Ask a question about your study material..."
    ),
    outputs=gr.Textbox(
        lines=8,
        label="AI Answer"
    ),
    title="Smart Student Learning Assistant",
    description="Ask questions and get answers from your uploaded study material."
)

demo.launch(share=True)
```
